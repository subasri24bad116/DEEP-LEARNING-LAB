```python
print("SUBASRI R 24BAD116")

# IMPORT LIBRARIES

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import os

print("TensorFlow Version:", tf.__version__)


# DATASET LOADING

(x_train, y_train), (_, _) = tf.keras.datasets.mnist.load_data()

print("Original Dataset Size:", x_train.shape)


# DISPLAY SAMPLE IMAGES

plt.figure(figsize=(8, 4))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.title("Digit " + str(y_train[i]))
    plt.axis("off")

plt.tight_layout()
plt.show()


# REDUCE DATASET SIZE FOR FAST TRAINING

# Use only 10000 images
x_train = x_train[:10000]

print("Training Dataset Size:", x_train.shape)


# DATA PREPROCESSING

# Convert to float
x_train = x_train.astype("float32")

# Normalize pixel values from [0,255] to [-1,1]
x_train = (x_train - 127.5) / 127.5

# Add channel dimension
x_train = x_train.reshape(
    x_train.shape[0],
    28,
    28,
    1
)

print("Preprocessed Shape:", x_train.shape)
print("Minimum Value:", x_train.min())
print("Maximum Value:", x_train.max())


# CREATE DATA PIPELINE

BATCH_SIZE = 256

train_dataset = (
    tf.data.Dataset
    .from_tensor_slices(x_train)
    .shuffle(10000)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

print("Data Pipeline Created Successfully")


# DEFINE GAN PARAMETERS

NOISE_DIM = 100
EPOCHS = 10

print("Noise Dimension:", NOISE_DIM)
print("Epochs:", EPOCHS)


# BUILD GENERATOR

generator = tf.keras.Sequential([

    tf.keras.layers.Dense(
        7 * 7 * 128,
        input_shape=(NOISE_DIM,)
    ),

    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.LeakyReLU(),

    tf.keras.layers.Reshape(
        (7, 7, 128)
    ),

    tf.keras.layers.Conv2DTranspose(
        64,
        kernel_size=5,
        strides=2,
        padding="same"
    ),

    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.LeakyReLU(),

    tf.keras.layers.Conv2DTranspose(
        32,
        kernel_size=5,
        strides=2,
        padding="same"
    ),

    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.LeakyReLU(),

    tf.keras.layers.Conv2D(
        1,
        kernel_size=5,
        padding="same",
        activation="tanh"
    )
])


# DISPLAY GENERATOR SUMMARY

print("\n========== GENERATOR SUMMARY ==========\n")

generator.summary()


# GENERATE IMAGE BEFORE TRAINING

random_noise = tf.random.normal(
    [1, NOISE_DIM]
)

before_training_image = generator(
    random_noise,
    training=False
)

plt.figure(figsize=(3, 3))

plt.imshow(
    before_training_image[0, :, :, 0],
    cmap="gray"
)

plt.title("Before Training")
plt.axis("off")
plt.show()


# BUILD DISCRIMINATOR

discriminator = tf.keras.Sequential([

    tf.keras.layers.Conv2D(
        64,
        kernel_size=5,
        strides=2,
        padding="same",
        input_shape=(28, 28, 1)
    ),

    tf.keras.layers.LeakyReLU(),
    tf.keras.layers.Dropout(0.3),

    tf.keras.layers.Conv2D(
        128,
        kernel_size=5,
        strides=2,
        padding="same"
    ),

    tf.keras.layers.LeakyReLU(),
    tf.keras.layers.Dropout(0.3),

    tf.keras.layers.Flatten(),

    tf.keras.layers.Dense(1)
])


# DISPLAY DISCRIMINATOR SUMMARY

print("\n========== DISCRIMINATOR SUMMARY ==========\n")

discriminator.summary()


# DEFINE LOSS FUNCTION

cross_entropy = tf.keras.losses.BinaryCrossentropy(
    from_logits=True
)


# DEFINE DISCRIMINATOR LOSS

def discriminator_loss(real_output, fake_output):

    real_loss = cross_entropy(
        tf.ones_like(real_output),
        real_output
    )

    fake_loss = cross_entropy(
        tf.zeros_like(fake_output),
        fake_output
    )

    return real_loss + fake_loss


# DEFINE GENERATOR LOSS

def generator_loss(fake_output):

    return cross_entropy(
        tf.ones_like(fake_output),
        fake_output
    )


# DEFINE OPTIMIZERS

generator_optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.0002,
    beta_1=0.5
)

discriminator_optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.0002,
    beta_1=0.5
)


# DEFINE TRAINING STEP

@tf.function
def train_step(images):

    noise = tf.random.normal(
        [tf.shape(images)[0], NOISE_DIM]
    )

    with tf.GradientTape() as gen_tape, \
         tf.GradientTape() as disc_tape:

        # Generate fake images
        generated_images = generator(
            noise,
            training=True
        )

        # Discriminator prediction
        real_output = discriminator(
            images,
            training=True
        )

        fake_output = discriminator(
            generated_images,
            training=True
        )

        # Calculate losses
        gen_loss = generator_loss(
            fake_output
        )

        disc_loss = discriminator_loss(
            real_output,
            fake_output
        )


    # Calculate gradients
    generator_gradients = gen_tape.gradient(
        gen_loss,
        generator.trainable_variables
    )

    discriminator_gradients = disc_tape.gradient(
        disc_loss,
        discriminator.trainable_variables
    )


    # Update weights
    generator_optimizer.apply_gradients(
        zip(
            generator_gradients,
            generator.trainable_variables
        )
    )

    discriminator_optimizer.apply_gradients(
        zip(
            discriminator_gradients,
            discriminator.trainable_variables
        )
    )

    return gen_loss, disc_loss


# CREATE FIXED RANDOM NOISE

seed = tf.random.normal(
    [16, NOISE_DIM]
)


# CREATE IMAGE DISPLAY FUNCTION

def generate_images(model, epoch):

    predictions = model(
        seed,
        training=False
    )

    plt.figure(figsize=(6, 6))

    for i in range(16):

        plt.subplot(4, 4, i + 1)

        image = (
            predictions[i, :, :, 0] * 0.5
            + 0.5
        )

        plt.imshow(
            image,
            cmap="gray"
        )

        plt.axis("off")

    plt.suptitle(
        "Generated Images - Epoch " + str(epoch)
    )

    plt.tight_layout()
    plt.show()


# INITIALIZE LOSS HISTORY

generator_losses = []
discriminator_losses = []


# GAN TRAINING

print("\n========== GAN TRAINING STARTED ==========\n")

for epoch in range(1, EPOCHS + 1):

    total_gen_loss = 0
    total_disc_loss = 0
    batch_count = 0

    for image_batch in train_dataset:

        gen_loss, disc_loss = train_step(
            image_batch
        )

        total_gen_loss += gen_loss
        total_disc_loss += disc_loss

        batch_count += 1


    # Calculate average losses

    avg_gen_loss = (
        total_gen_loss / batch_count
    )

    avg_disc_loss = (
        total_disc_loss / batch_count
    )


    # Store losses

    generator_losses.append(
        float(avg_gen_loss)
    )

    discriminator_losses.append(
        float(avg_disc_loss)
    )


    # Display training progress

    print(
        f"Epoch {epoch}/{EPOCHS} | "
        f"Generator Loss: {avg_gen_loss:.4f} | "
        f"Discriminator Loss: {avg_disc_loss:.4f}"
    )


    # Display images only at selected epochs

    if epoch == 1 or epoch == 5 or epoch == 10:

        generate_images(
            generator,
            epoch
        )


# FINAL SYNTHETIC IMAGES

print("\n========== FINAL SYNTHETIC IMAGES ==========\n")

final_noise = tf.random.normal(
    [16, NOISE_DIM]
)

final_images = generator(
    final_noise,
    training=False
)

plt.figure(figsize=(6, 6))

for i in range(16):

    plt.subplot(4, 4, i + 1)

    image = (
        final_images[i, :, :, 0] * 0.5
        + 0.5
    )

    plt.imshow(
        image,
        cmap="gray"
    )

    plt.axis("off")

plt.suptitle("Final Synthetic Images")
plt.tight_layout()
plt.show()


# LOSS GRAPH

plt.figure(figsize=(8, 4))

plt.plot(
    generator_losses,
    label="Generator Loss"
)

plt.plot(
    discriminator_losses,
    label="Discriminator Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("GAN Training Loss")
plt.legend()
plt.grid()

plt.show()


# REAL VS SYNTHETIC IMAGES

plt.figure(figsize=(10, 4))

# Real images
for i in range(5):

    plt.subplot(2, 5, i + 1)

    plt.imshow(
        x_train[i, :, :, 0] * 0.5 + 0.5,
        cmap="gray"
    )

    plt.title("Real")
    plt.axis("off")


# Synthetic images
for i in range(5):

    plt.subplot(2, 5, i + 6)

    plt.imshow(
        final_images[i, :, :, 0] * 0.5 + 0.5,
        cmap="gray"
    )

    plt.title("Synthetic")
    plt.axis("off")

plt.tight_layout()
plt.show()


# SAVE GENERATED IMAGES

os.makedirs(
    "generated_images",
    exist_ok=True
)

for i in range(16):

    image = (
        final_images[i, :, :, 0] * 0.5
        + 0.5
    )

    plt.imsave(
        f"generated_images/generated_{i+1}.png",
        image.numpy(),
        cmap="gray"
    )

print("Generated images saved successfully.")
print("GAN experiment completed successfully.")
```
